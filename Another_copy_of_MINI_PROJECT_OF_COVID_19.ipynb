```python
# get 10 linearily spaced points in the interval[0,10,20]

x=np.linspace(0,10,20)
y=x**2
print(x)
print(y)
```

```python
#create a figure/caves of desired size
plt.figure(figsize=(7,6))

#plot values;with a color 'red'
plt.plot(x,y,'r')

#give labels to the axes
plt.xlabel('x axis')
plt.ylabel('y axis')

#give a title to the plot
plt.title(r"Plot of $y=x^2$")
plt.show()
```

```python
people=['charlotte','Charli','amelia','sohia','william']
age=[21,23,35,45,29]
weight=[55,58,45,49,60]
height=[160,135,170,165,173]
```

```python
experience=[1,2,3,4,5]

workers_salary = [6500,9280,12050,13200,16672]
doctors_salary = [9020,12873,15725,18000,19700]
```

```python
plt.bar(experience,workers_salary,color="b")

plt.title("Salary of workers")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```

```python
plt.bar(experience,workers_salary,color="g",label="workers salary")
plt.plot(experience,doctors_salary,color="r",label="doctors salary")

plt.title("salary of workers of doctors by their experiences")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.legend()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
Covid=sns.load_dataset("tips")
```

```python
Covid
```

```python
#tips
#stripplot
sns.stripplot(y='tip',data=Covid,jitter=True)
plt.ylabel("tips.($)")
```

```python
sns.stripplot(y='day',data=Covid,jitter=False)
plt.ylabel("Covid,($)")
```

```python
#changing the orientation n
sns.swarmplot(x='day',y='tip',data=Covid)
plt.ylabel("Covid.($)")
```

```python
sns.swarmplot(x='day',y='tip',data=Covid,hue='sex')
plt.ylabel("Covid.($)")
```

```python
sns.swarmplot(x='day',y='tip',data=Covid,hue='sex',orient='h')
plt.ylabel("Covid.($)")
```

```python
sns.swarmplot(x='day',y='tip',data=Covid)
plt.ylabel("Covid.($)")
```

```python
sns.stripplot(y='tip',x='day',data=Covid)
plt.ylabel("Covid,($)")
```

```python
sns.violinplot(x='day',y='tip',data=Covid)
plt.ylabel("Covid.($)")
```

```python
sns.violinplot(x='day',y='tip',data=Covid,inner=None,color='lightblue')
plt.ylabel("Covid.($)")
```
